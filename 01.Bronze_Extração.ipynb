```python
display(spark.sql("DESCRIBE TABLE samples.nyctaxi.trips"))
```

```python
df_raw = spark.read.table("samples.nyctaxi.trips")

df_raw.printSchema()

print(f"Total de linhas: {df_raw.count()}")
```

```python
df_bronze = spark.read.table("samples.nyctaxi.trips")

df_bronze.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.bronze_nyctaxi")

print("Tabela bronze_nyctaxi criada no catálogo Unity Catalog.")
```

```python


(df_raw.write
    .format("delta")              # formato Delta Lake
    .mode("overwrite")             # se já existir uma tabela ali, substitui
    .option("overwriteSchema", "true")  # permite que o schema também mude, não só os dados
    .save("/Volumes/workspace/default/lakehouse/bronze/nyctaxi"))  # caminho dentro do Volume

print("Bronze gravado com sucesso.")
```
